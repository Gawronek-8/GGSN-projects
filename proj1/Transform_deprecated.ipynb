```python
import os
import shutil
import csv
from utils import ROOT_PATH
```

```python
def dataset_to_csv(src_name, output_name): # Function for parsing datasets from OpenML

    data_path = ROOT_PATH / "data"
    headers = []

    with open(data_path / f"{output_name}.csv", 'w') as csvfile:
        with open(data_path / f"{src_name}", 'r') as f:
            for line in f:
                if '@ATTRIBUTE' not in line.strip():
                    continue
                try:
                    headers.append(line.strip().split(' ')[1])
                except IndexError:
                    print(f"Error while parsing header {line.strip()}")

        writer = csv.writer(csvfile, delimiter=',')
        writer.writerow(headers)

        with open(data_path / f"{src_name}", 'r') as f:
            data_tag_encountered = False
            for line in f:
                if line.strip() == '@DATA':
                    data_tag_encountered = True

                if data_tag_encountered and line.strip() != '@DATA':
                    data_row = line.strip().split(',')
                    writer.writerow(data_row)
                else:
                    continue
```

```python
dataset_to_csv("dataset_", "sepsa_dataset")
```
